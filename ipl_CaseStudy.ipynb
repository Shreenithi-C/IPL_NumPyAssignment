```python
import numpy as np

# Load dataset once (skip header row)
data = np.genfromtxt("deliveries.csv", delimiter=",", dtype=str, skip_header=1)

# Quick check: print shape and first row
print("Shape:", data.shape)
print("First row:", data[0])
```

```python
#Task 1 - Total runs per match 

# Extract relevant columns
match_ids = data[:, 0].astype(int)        
batsman_runs = data[:, 9].astype(int)   

# Map each match_id to a unique index
unique_matches, match_index = np.unique(match_ids, return_inverse=True)

# Aggregate runs per match
total_runs = np.bincount(match_index, weights=batsman_runs)

# Combine results
results_task1 = np.column_stack((unique_matches, total_runs))

# Display first 10 results
print("match_id | total_runs")
print(results_task1[:10])
```

```python
#Task 2 - Top 5 batters

# Extract relevant columns
batters = data[:, 6]                     # batter column
batsman_runs = data[:, 9].astype(int)    # batsman_runs column

# Get unique batters and map each row to a batter index
unique_batters, batter_index = np.unique(batters, return_inverse=True)

# Aggregate runs per batter
total_runs_per_batter = np.bincount(batter_index, weights=batsman_runs)

# Sort batters by total runs (descending)
sorted_indices = np.argsort(total_runs_per_batter)[::-1]

# Select top 5
top5_indices = sorted_indices[:5]
top5_batters = unique_batters[top5_indices]
top5_runs = total_runs_per_batter[top5_indices]

# Combine results
results_task2 = np.column_stack((top5_batters, top5_runs.astype(int)))

print("Top 5 Batters (player_name | total_runs)")
print(results_task2)
```

```python
#Task 3 - Strike rate of batters

# Extract relevant columns
batters = data[:, 6]                     # batter column
batsman_runs = data[:, 9].astype(int)    # batsman_runs column

# Each row = one ball faced, so count balls per batter
unique_batters, batter_index = np.unique(batters, return_inverse=True)

# Total runs per batter
total_runs_per_batter = np.bincount(batter_index, weights=batsman_runs)

# Balls faced per batter (just count occurrences)
balls_faced_per_batter = np.bincount(batter_index)

# Calculate strike rate: (runs / balls) * 100
strike_rates = (total_runs_per_batter / balls_faced_per_batter) * 100

# Combine results
results_task3 = np.column_stack((unique_batters, strike_rates.round(2)))

# Display first 10 batters with strike rates
print("batter | strike_rate")
print(results_task3[:10])
```

```python
#Task 4 - Economy Rate of bowlers

# Extract relevant columns
bowlers = data[:, 7]                     # bowler column
batsman_runs = data[:, 9].astype(int)    # runs scored on that ball

# Map each row to a bowler index
unique_bowlers, bowler_index = np.unique(bowlers, return_inverse=True)

# Runs conceded per bowler
runs_conceded = np.bincount(bowler_index, weights=batsman_runs)

# Balls bowled per bowler (each row = one ball bowled)
balls_bowled = np.bincount(bowler_index)

# Convert balls to overs (balls / 6)
overs_bowled = balls_bowled / 6

# Economy rate = runs conceded / overs bowled
economy_rates = runs_conceded / overs_bowled

# Combine results
results_task4 = np.column_stack((unique_bowlers, economy_rates.round(2)))

# Display first 10 bowlers with economy rates
print("bowler | economy_rate")
print(results_task4[:10])
```

```python
#Task 5 - Runs per over 

# Extract relevant columns
overs = data[:, 4].astype(int)          
batsman_runs = data[:, 9].astype(int)    

# Map each row to an over index (1–20)
unique_overs, over_index = np.unique(overs, return_inverse=True)

# Total runs per over
runs_per_over = np.bincount(over_index, weights=batsman_runs)

# Balls bowled per over (each row = one ball)
balls_per_over = np.bincount(over_index)

# Average runs per over = total runs / number of balls * 6
# (since each over has 6 balls, we normalize to per-over average)
avg_runs_per_over = (runs_per_over / balls_per_over) * 6

# Ensure output is size 20 (overs 1–20)
results_task5 = np.column_stack((unique_overs, avg_runs_per_over.round(2)))

print("over | average_runs")
print(results_task5)
```

```python
#Task 6 - Boundary Analysis 

# Extract relevant columns
batsman_runs = data[:, 9].astype(int)    
batting_team = data[:, 2]                

# Count total number of 4s and 6s
total_fours = np.sum(batsman_runs == 4)
total_sixes = np.sum(batsman_runs == 6)

print("Total 4s:", total_fours)
print("Total 6s:", total_sixes)

# Bonus: Identify which team has hit the most boundaries
# Mask for boundaries (4s or 6s)
boundary_mask = (batsman_runs == 4) | (batsman_runs == 6)

# Teams corresponding to boundary balls
boundary_teams = batting_team[boundary_mask]

# Map each boundary to a team index
unique_teams, team_index = np.unique(boundary_teams, return_inverse=True)

# Count boundaries per team
boundaries_per_team = np.bincount(team_index)

# Find team with most boundaries
max_team_index = np.argmax(boundaries_per_team)
team_with_most_boundaries = unique_teams[max_team_index]
max_boundaries = boundaries_per_team[max_team_index]

print("Team with most boundaries:", team_with_most_boundaries, "(", max_boundaries, ")")
```

```python
# Task 7 - Death overs Analysis 

# Extract columns
overs = data[:, 4].astype(int)          
batsman_runs = data[:, 9].astype(int)   
batting_team = data[:, 2]                

# Mask for death overs 
death_mask = (overs >= 16) & (overs <= 19)

print("Death overs rows:", np.sum(death_mask))  

# Total runs in death overs
total_runs_death_overs = np.sum(batsman_runs[death_mask])
print("Total runs in death overs:", total_runs_death_overs)

# Team-wise runs in death overs
death_teams = batting_team[death_mask]
death_runs = batsman_runs[death_mask]

if death_runs.size == 0:
    print("No death overs data found.")
else:
    unique_teams, team_index = np.unique(death_teams, return_inverse=True)
    runs_per_team_death = np.bincount(team_index, weights=death_runs)

    max_team_index = np.argmax(runs_per_team_death)
    team_with_max_death_runs = unique_teams[max_team_index]
    max_death_runs = runs_per_team_death[max_team_index]

    print("Team with highest runs in death overs:", team_with_max_death_runs, "(", int(max_death_runs), ")")
```

```python
# Task 8 - Highest Scoring Match 

# Extract relevant columns
match_ids = data[:, 0].astype(int)       
batsman_runs = data[:, 9].astype(int)   

# Map each row to a match index
unique_matches, match_index = np.unique(match_ids, return_inverse=True)

# Aggregate runs per match
total_runs_per_match = np.bincount(match_index, weights=batsman_runs)

# Find match with highest total runs
max_match_index = np.argmax(total_runs_per_match)
highest_match_id = unique_matches[max_match_index]
highest_total_runs = total_runs_per_match[max_match_index]

print("Highest scoring match:", highest_match_id, "with", int(highest_total_runs), "runs")
```

```python
#Task 9 - Runs per Team per Match

# Extract relevant columns
match_ids = data[:, 0].astype(int)       
batting_team = data[:, 2]                
batsman_runs = data[:, 9].astype(int)    

# Create composite key: (match_id, team)
# We can combine into strings like "matchid_team"
composite_keys = np.char.add(match_ids.astype(str), "_" + batting_team)

# Map each row to a composite index
unique_keys, key_index = np.unique(composite_keys, return_inverse=True)

# Aggregate runs per (match, team)
runs_per_key = np.bincount(key_index, weights=batsman_runs)

# Split composite keys back into match_id and team
split_keys = np.char.split(unique_keys, sep="_")
split_keys = np.array([list(k) for k in split_keys])  # convert list of lists to array

# Combine results
results_task9 = np.column_stack((split_keys[:,0], split_keys[:,1], runs_per_key.astype(int)))

# Display first 10 results
print("match_id | team | total_runs")
print(results_task9[:10])
```

```python
#Task 10 - Match Winner Approximation

# Extract relevant columns
match_ids = data[:, 0].astype(int)       
batting_team = data[:, 2]                
batsman_runs = data[:, 9].astype(int)   

# Create composite key: (match_id, team)
composite_keys = np.char.add(match_ids.astype(str), "_" + batting_team)

# Map each row to composite index
unique_keys, key_index = np.unique(composite_keys, return_inverse=True)

# Aggregate runs per (match, team)
runs_per_key = np.bincount(key_index, weights=batsman_runs)

# Split composite keys back into match_id and team
split_keys = np.char.split(unique_keys, sep="_")
split_keys = np.array([list(k) for k in split_keys])

# Combine results into structured array
match_team_runs = np.column_stack((split_keys[:,0], split_keys[:,1], runs_per_key.astype(int)))

# Now determine winner per match
winners = []
unique_matches = np.unique(match_ids)

for match in unique_matches:
    # Filter rows for this match
    mask = match_team_runs[:,0] == str(match)
    teams = match_team_runs[mask][:,1]
    runs = match_team_runs[mask][:,2].astype(int)
    
    if runs.size == 2:  # only consider matches with 2 teams
        winner_team = teams[np.argmax(runs)]
        winners.append([match, winner_team])

results_task10 = np.array(winners)

print("match_id | winner_team")
print(results_task10[:10])  # show first 10 winners
```

```python
#Task 11 - Match Scorecard Generation 

# Extract relevant columns
match_ids = data[:, 0].astype(int)       
batting_team = data[:, 2]                
batsman_runs = data[:, 9].astype(int)    

# Create composite key: (match_id, team)
composite_keys = np.char.add(match_ids.astype(str), "_" + batting_team)

# Map each row to composite index
unique_keys, key_index = np.unique(composite_keys, return_inverse=True)

# Aggregate runs per (match, team)
runs_per_key = np.bincount(key_index, weights=batsman_runs)

# Split composite keys back into match_id and team
split_keys = np.char.split(unique_keys, sep="_")
split_keys = np.array([list(k) for k in split_keys])

# Combine results into structured array
match_team_runs = np.column_stack((split_keys[:,0], split_keys[:,1], runs_per_key.astype(int)))

# Generate scorecards
unique_matches = np.unique(match_ids)

for i, match in enumerate(unique_matches, start=1):
    mask = match_team_runs[:,0] == str(match)
    teams = match_team_runs[mask][:,1]
    runs = match_team_runs[mask][:,2].astype(int)
    
    print(f"Match {i}:")
    for team, run in zip(teams, runs):
        print(f"  {team}: {run} runs")
    print()
```
